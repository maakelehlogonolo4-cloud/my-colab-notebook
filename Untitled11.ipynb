```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(15,10))
sns.boxplot(x='Placement', y='Academic_Performance', data=df)
plt.title('Placement vs Academic_Performance')
plt.xlabel('Placement')
plt.ylabel('Academic_Performance')
plt.show()
```

```python
sns.histplot(df['Placement'], kde=True)
plt.title('Histogram of Placement')
plt.xlabel('Placement')
plt.ylabel('Frequency')
plt.show()
```

```python

plt.figure(figsize=(10,6))
sns.scatterplot(x='Placement', y='Academic_Performance', data=df)
plt.title('Scatter Plot of Placement vs Academic_Performance')
plt.xlabel('Placement')
plt.ylabel('Academic_Performance')
plt.show()
```

```python
#BOX PLOT FOR CGPA VS INTERNSHIP_EXPERIENCE
plt.figure(figsize=(20,15))
plt.subplot(3,3,1)
sns.boxplot(x='Internship_Experience', y='CGPA', data=df)
plt.title('Internship_Experience vs CGPA')
plt.xlabel('Internship_Experience')
plt.ylabel('CGPA')
plt.show()
```

```python
#BAR PLOT FOR PLACEMENT COUNT
plt.figure(figsize=(20,15))
sns.countplot(x='Placement', data=df)
plt.title('Placement Status By Count Of Student')
plt.xlabel('Placement Status')
plt.ylabel('Count')
plt.show()
```

```python
#BAR PLOT FOR COMMUNICATION_SKILLS DISTRIBUTION
plt.figure(figsize=(20,15))
sns.countplot(x='Communication_Skills', data=df)
plt.title('Communication_Skills Distribution')
plt.xlabel('Communication_Skills')
plt.ylabel('Count')
plt.show()
```
