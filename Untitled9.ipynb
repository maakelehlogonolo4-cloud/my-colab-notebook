```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Box Plot of 'price' and 'room_type'
plt.figure(figsize=(10,6))
sns.boxplot(x='price', y= 'room_type', data=df)
plt.title('Price Distribution By Room Type')
plt.xlabel('Price')
plt.ylabel('Room Type')
plt.show()
```

```python
#Histogram of 'price'
sns.histplot(df['price']. dropna(), bins=20, kde=True)
plt.title('Distribution of Price')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```python
#Scatterplot of 'accomodates' vs 'price'
sns.scatterplot(x='accommodates', y='price', data=df)
plt.title('Accomodates vs Price')
plt.xlabel('Accomodates')
plt.ylabel('Price')
plt.show()
```
